```python
def build_incidence_matrix(df, num_users, num_items, num_times):
    M = len(df)
    N = int(num_users + num_items + num_times)

    rows, cols, vals = [], [], []
    for local_idx, (_, row) in enumerate(tqdm(df.iterrows(), total=M)):
        u_node = row[0]
        i_node = num_users + row[1]
        t_node = num_users + num_items + row[2]

        rows.extend([u_node, i_node, t_node])
        cols.extend([local_idx, local_idx, local_idx])  # 여기에서 local_idx 사용!
        vals.extend([1, 1, 1])

    H = sp.coo_matrix((vals, (rows, cols)), shape=(N, M)).tocsr()
    return H
```

```python
def build_weighted_incidence_clique(H):
    """
    H: scipy.sparse matrix (N x M), incidence matrix (0/1)
    ratings: numpy array (M,), interaction ratings
    return: A_weighted (N x N) sparse matrix, weighted clique expansion
    """
    A = H.dot(H.T)
    
    # 대각선은 0으로 (자기 자신 연결 제외)
    A.setdiag(0)
    A.eliminate_zeros()
    
    return A
```

```python
H = build_incidence_matrix(X_train, num_users, num_items, num_times)
```

```python
A = build_weighted_incidence_clique(H)
```

```python
def symmetric_normalize_adj(A):
    """
    A: scipy.sparse matrix (square, symmetric or not)
    return: normalized A_hat = D^{-1/2} A D^{-1/2}
    """
    # degree 벡터 D_ii = sum_j A_ij
    degrees = np.array(A.sum(axis=1)).flatten()
    
    # 0인 원소에 대해서는 1로 처리 (0으로 나누는 것 방지)
    degrees[degrees == 0] = 1
    
    # D^{-1/2}
    d_inv_sqrt = 1.0 / np.sqrt(degrees)
    
    # 대각행렬로 변환
    D_inv_sqrt = sp.diags(d_inv_sqrt)
    
    # 정규화된 행렬 계산
    A_normalized = D_inv_sqrt.dot(A).dot(D_inv_sqrt)
    return A_normalized
```

```python
def sparse_to_pandas_df(sparse_mat):
    """
    sparse_mat: scipy.sparse matrix (예: csr_matrix)
    return: pandas DataFrame with columns ['i', 'j', 'weight']
    """
    coo = sparse_mat.tocoo()  # COO 포맷으로 변환
    df = pd.DataFrame({
        'i': coo.row,
        'j': coo.col,
        'weight': coo.data
    })
    return df

# 사용 예시
df_edges = sparse_to_pandas_df(symmetric_normalize_adj(A))
print(df_edges.head())
```
