```python
import numpy as np
import pandas as pd
import os
from tqdm import tqdm
import time
import scipy
from scipy.sparse import coo_matrix, csr_matrix
import warnings
warnings.filterwarnings('ignore')


def hg_construction(df_train, num_entities, N):


    e = int(df_train.shape[0])

    hypergraph = np.zeros((int(np.sum(num_entities)), e))
    sparse_hyper = np.zeros((N*e, 2));

    for i in tqdm(range(e)):
        for n in range(N):
            idx_n = df_train.to_numpy()[i,n]
            sparse_hyper[N*i+n,0] = idx_n + np.sum(num_entities[:n])
            sparse_hyper[N*i+n,1] = i


    sparse_hyper = sparse_hyper.astype(np.int64)
    hypergraph = coo_matrix((np.ones_like(sparse_hyper[:,0]), (sparse_hyper[:,0], sparse_hyper[:,1])));
    hypergraph = csr_matrix(hypergraph)

    return hypergraph

def clique_expansion(hypergraph):
    sim = hypergraph.dot(hypergraph.T).todense()
    sim = sim - np.diag(np.diagonal(sim));
    sum_sim = np.asarray(np.sum(sim, axis=1)).squeeze()**(-1/2)
    sum_sim[sum_sim == np.inf] = 0
    sum_sim = np.diag(sum_sim)
    sim_normalized = sum_sim*(sim)*sum_sim;

    return sim_normalized
```

```python
data_name = 'yelp'
path = './'
df_train = pd.read_csv(os.path.join(path,'yelp_train.tsv'), header=None, sep='\t')
df_valid = pd.read_csv(os.path.join(path,'yelp_valid.tsv'), header=None, sep='\t')
df_test = pd.read_csv(os.path.join(path,'yelp_test.tsv'), header=None, sep='\t')

N = df_train.shape[1]

num_entities = np.zeros((N,))
for n in range(N):
    num_entities[n] = np.max([df_train[n].to_numpy().max(), df_valid[n].to_numpy().max(), df_test[n].to_numpy().max()])+1


hypergraph = hg_construction(df_train, num_entities, N);
# clique_expansion_graph = clique_expansion(hypergraph)

# pd.DataFrame(clique_expansion_graph).to_csv(os.path.join(path, './returna_'+data_name+'.txt'), header=None, sep='\t', index=False)
```

```python
def hg_construction3(df_train, num_entities, N):
    e = df_train.shape[0]  # Number of edges
    num_rows, num_cols = df_train.shape

    # Compute the offsets for each entity group
    offsets = np.cumsum(np.concatenate((np.zeros((1,)), (num_entities[:-1]))))
    print(offsets)
    # Convert the DataFrame to a NumPy array for fast indexing
    data_array = df_train.to_numpy()

    # Repeat edge indices for each node (0, 1, ..., N-1 for every edge)
    edge_indices = np.repeat(np.arange(e), N)

    expanded_offsets = np.tile(offsets, e)
    print(expanded_offsets)

    # Compute the indices for the nodes in the hypergraph
    node_indices = data_array.ravel() + expanded_offsets

    print(node_indices.shape)
    print(node_indices)
    # Combine the node and edge indices into the sparse hypergraph representation
    sparse_hyper = np.column_stack((node_indices, edge_indices)).astype(np.int64)

    return sparse_hyper
```

```python
import time

start = time.time()
hypergraph3 = hg_construction3(df_train, num_entities, N);
end = time.time()

print(f"{end - start:.5f} sec")
```
