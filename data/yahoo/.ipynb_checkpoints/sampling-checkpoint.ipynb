```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
np.random.seed(10)
a1 = a[:a.shape[0]//16]
b1 = b[:b.shape[0]//16]

a2 = a[:a.shape[0]//4]
b2 = b[:b.shape[0]//16]

a3 = a[:a.shape[0]//4]
b3 = b[:b.shape[0]//4]

a4 = a[:a.shape[0]]
b4 = b[:b.shape[0]//4]


# np.random.seed(10)
# a1 = np.random.choice(a, a.shape[0]//16, replace=False)
# b1 = np.random.choice(b, b.shape[0]//16, replace=False)

# a2 = np.random.choice(a, a.shape[0]//4, replace=False)
# b2 = np.random.choice(b, b.shape[0]//16, replace=False)

# a3 = np.random.choice(a, a.shape[0]//4, replace=False)
# b3 = np.random.choice(b, b.shape[0]//4, replace=False)

# a4 = np.random.choice(a, a.shape[0], replace=False)
# b4 = np.random.choice(b, b.shape[0]//4, replace=False)

# a5 = np.random.choice(a, a.shape[0], replace=False)
# b5 = np.random.choice(b, b.shape[0], replace=False)
```

```python
filtered_df = df[df[0].isin(a1)]
filtered_df = filtered_df[filtered_df[1].isin(b1)]
filtered_df = filtered_df[[0,1,2]]
print(filtered_df.shape[0])

a_key_val_dict = {}
b_key_val_dict = {}

for i in range(len(a1)):
    a_key_val_dict[a1[i]] = i

for i in range(len(b1)):
    b_key_val_dict[b1[i]] = i


filtered_df[0] = filtered_df[0].apply(lambda x: a_key_val_dict[x])
filtered_df[1] = filtered_df[1].apply(lambda x: b_key_val_dict[x])

print(filtered_df[0].max()+1)
print(filtered_df[1].max()+1)
print(filtered_df[2].max()+1)

X_train1, X_test1 = train_test_split(filtered_df, test_size=0.4, random_state=42)
X_test1, X_valid1 = train_test_split(X_test1, test_size=0.5, random_state=42)

X_train1.to_csv('./yahoo_train1.tsv', sep='\t', header=None, index=False)
X_valid1.to_csv('./yahoo_valid1.tsv', sep='\t', header=None, index=False)
X_test1.to_csv('./yahoo_test1.tsv', sep='\t', header=None, index=False)
```
