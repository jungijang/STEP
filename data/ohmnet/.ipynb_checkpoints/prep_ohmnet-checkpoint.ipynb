```python
def build_incidence_matrix(df, num_users, num_items, num_times):
    df = df.reset_index(drop=True)  # col 매칭 안전하게
    M = len(df)
    N = num_users + num_items + num_times

    u_nodes = df.iloc[:,0].to_numpy(np.int64)
    i_nodes = num_users + df.iloc[:,1].to_numpy(np.int64)
    t_nodes = num_users + num_items + df.iloc[:,2].to_numpy(np.int64)

    # (M, 3) 배열로 쌓아서 flatten → [u0,i0,t0, u1,i1,t1, ...]
    rows = np.stack([u_nodes, i_nodes, t_nodes], axis=1).ravel()

    # 각 interaction마다 col index 동일
    cols = np.repeat(np.arange(M, dtype=np.int64), 3)

    vals = np.ones(3*M, dtype=np.int8)

    H = sp.coo_matrix((vals, (rows, cols)), shape=(N, M)).tocsr()
    return H
```

```python
def build_weighted_incidence_clique(H):
    """
    H: scipy.sparse matrix (N x M), incidence matrix (0/1)
    ratings: numpy array (M,), interaction ratings
    return: A_weighted (N x N) sparse matrix, weighted clique expansion
    """
    A = H.dot(H.T)
    
    # 대각선은 0으로 (자기 자신 연결 제외)
    A.setdiag(0)
    A.eliminate_zeros()
    
    return A
```

```python
def csr_memory_in_mb(A):
    if not isinstance(A, sp.csr_matrix):
        A = A.tocsr()
    mem_bytes = A.data.nbytes + A.indices.nbytes + A.indptr.nbytes
    return mem_bytes / (1024 * 1024)

print(f"CSR size: {csr_memory_in_mb(H):.2f} MB")
print(f"CSR size: {csr_memory_in_mb(A_norm):.2f} MB")
```

```python
def symmetric_normalize_adj(A):
    """
    A: scipy.sparse matrix (square, symmetric or not)
    return: normalized A_hat = D^{-1/2} A D^{-1/2}
    """
    # degree 벡터 D_ii = sum_j A_ij
    degrees = np.array(A.sum(axis=1)).flatten()
    
    # 0인 원소에 대해서는 1로 처리 (0으로 나누는 것 방지)
    degrees[degrees == 0] = 1
    
    # D^{-1/2}
    d_inv_sqrt = 1.0 / np.sqrt(degrees)
    
    # 대각행렬로 변환
    D_inv_sqrt = sp.diags(d_inv_sqrt)
    
    # 정규화된 행렬 계산
    A_normalized = D_inv_sqrt.dot(A).dot(D_inv_sqrt)
    return A_normalized
```

```python
def sparse_to_pandas_df(sparse_mat):
    """
    sparse_mat: scipy.sparse matrix (예: csr_matrix)
    return: pandas DataFrame with columns ['i', 'j', 'weight']
    """
    coo = sparse_mat.tocoo()  # COO 포맷으로 변환
    df = pd.DataFrame({
        'i': coo.row,
        'j': coo.col,
        'weight': coo.data
    })
    return df

# 사용 예시
df_edges = sparse_to_pandas_df(symmetric_normalize_adj(A))
print(df_edges.head())
```
